```python

# Simple Regression Tree Implementation

#Algorithm
#Select a feature
#1.Try possible split points
#2.Calculate MSE (Mean Squared Error) for each split
#3.Select the split with the lowest MSE
#4.Calculate the prediction in each leaf using the mean
#5.Predict for a new value

# Dataset
X = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

# New value to predict
test = 3.5


# Step 1: Calculate MSE
def mse(values):
    mean = sum(values) / len(values)
    error = 0

    for value in values:
        error += (value - mean) ** 2

    return error / len(values)


# Step 2: Try all possible split points
best_mse = float("inf")
best_split = None

for split in [1.5, 2.5, 3.5, 4.5]:

    left = []
    right = []

    for i in range(len(X)):
        if X[i] <= split:
            left.append(y[i])
        else:
            right.append(y[i])

    # Calculate weighted MSE
    total = len(y)

    weighted_mse = (
        len(left) / total * mse(left)
        + len(right) / total * mse(right)
    )

    print("Split:", split)
    print("Left:", left)
    print("Right:", right)
    print("MSE:", weighted_mse)
    print()

    # Step 3: Select lowest MSE
    if weighted_mse < best_mse:
        best_mse = weighted_mse
        best_split = split


# Step 4: Create final groups
left = []
right = []

for i in range(len(X)):
    if X[i] <= best_split:
        left.append(y[i])
    else:
        right.append(y[i])


# Step 5: Calculate leaf predictions
left_prediction = sum(left) / len(left)
right_prediction = sum(right) / len(right)

print("Best Split:", best_split)
print("Left Leaf Prediction:", left_prediction)
print("Right Leaf Prediction:", right_prediction)


# Step 6: Predict new value
if test <= best_split:
    prediction = left_prediction
else:
    prediction = right_prediction

print("Prediction for", test, "=", prediction)
```
